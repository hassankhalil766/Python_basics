```python
a=5
b=10
sum=a+b
print("sum of a and b {}".format(sum))#normal formatting
print("sum of {1} and {0} is {2}".format(a,b,sum))# index based formatting
print("value of {a}and {b} is {sum}".format(a=5,b=10,sum=a+b))# value based formatting
print(f"sum of {a} and {b} is {sum}")
```

```python
marks=[89,99,100,1.95,20,12]
print(marks[1])
print(marks)
print(len(marks))
print(type(marks))
marks[2]=80
print(marks)
marks[3:6]
marks.append(80)
print(marks)
marks.insert(3,120)
print(marks)
marks.sort()
print(marks)
marks.sort(reverse=True)
print(marks)
marks.reverse()
print(marks)
```

```python
marks= [1,2,3,4,5,6,8]
idx=0
for val in marks:
    if(val==6):
        print(f"index found at {idx}")
        break
    idx+=1
```

```python
tup= (1,2,2,3,3,4,5)
print(tup)
print(len(tup))
print(type(tup))
print(tup[0:3])
sum=0
for val in tup:
    print(val)
    sum+=val
print(sum)
tup.index(2)
tup.count(3)
```
